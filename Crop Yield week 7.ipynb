```python
import pandas as pd

df = pd.read_csv("C:/Users/ELCOT/Downloads/crop_yield.csv")

print(df.columns.tolist())
```

```python
df = pd.get_dummies(df, columns=["Crop"], dtype=int)
```

```python
df.head()
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
print(df["Season"].unique())
```

```python
df["Season"] = df["Season"].str.strip()
```

```python
from sklearn.preprocessing import OrdinalEncoder

oe = OrdinalEncoder(
    categories=[["Whole Year", "Kharif", "Rabi", "Summer", "Winter", "Autumn"]]
)

df["Season_Encoded"] = oe.fit_transform(df[["Season"]])
```

```python
print(df[["Season", "Season_Encoded"]].head())
```

```python
col = df["Yield"].dropna()
```

```python
print("Mean:", col.mean())
print("Median:", col.median())
print("Mode:", col.mode().tolist())
```

```python
print("Standard Deviation:", col.std())
print("Variance:", col.var())
```

```python
print("Minimum:", col.min())
print("Maximum:", col.max())
print("Range:", col.max() - col.min())
```

```python
Q1 = col.quantile(0.25)
Q2 = col.quantile(0.50)
Q3 = col.quantile(0.75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q2:", Q2)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
print("Skewness:", col.skew())
```

```python
print("Kurtosis:", col.kurt())
```

```python
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
```

```python
outliers = col[(col < lower) | (col > upper)]
```

```python
print(outliers)
```

```python
print("Lower Limit:", lower)
print("Upper Limit:", upper)
print("Outliers:", outliers.tolist())
print("Number of Outliers:", len(outliers))
```

```python
if col.mean() != 0:
    cv = (col.std() / col.mean()) * 100
    print("Coefficient of Variation (CV):", cv, "%")
else:
    print("CV cannot be calculated because the mean is zero.")
```

```python
mean = col.mean()
median = col.median()

if mean > median:
    print("Distribution may be right-skewed")
elif mean < median:
    print("Distribution may be left-skewed")
else:
    print("Mean and median are equal")
```
